```python
# check null values presence >50% 
missing_percent = hma_p.isna().mean().sort_values(ascending=False)
print(missing_percent.head(40))
```

```python
# drop small number of null values (0.001%)
hma_p = hma_p.dropna(subset="status")

# drop unecessary columns 
hma_p = hma_p.drop(columns=['subrecipientTribalIndicator', 'subrecipientAdminCostAmt', 'srmcObligatedAmt', 'recipientAdminCostAmt','dateInitiallyApproved', 'initialObligationDate', 'initialObligationAmount', 'id', 'countyCode'])

hma_p.isna().sum()
```

```python
hma_p.shape
```

```python
# format projectCounties - lowercase 
hma_p['projectCounties'] = hma_p['projectCounties'].str.capitalize()
print(hma_p['projectCounties'])
```

```python
# Match County and projectCounties

def extract_clean_county(value):
    if isinstance(value, str):
        # Handle known messy entries
        if "Multiple" in value or "Various" in value:
            return None
        # Extract first entry
        first = value.split(',')[0]
        # Remove words like "County" or "Parish", clean whitespace
        return (
            first.replace("County", "")
                 .replace("Parish", "")
                 .strip()
        )
    return None

mask = hma_p['county'].isna() & hma_p['projectCounties'].notna()

hma_p.loc[mask, 'county'] = hma_p.loc[mask, 'projectCounties'].apply(extract_clean_county)
print("Remaining missing values in `county`:", hma_p['county'].isna().sum())
```

```python
hma_p.isna().sum()
```

```python
# Fill projectCounties with cleaned county names where it's missing
mask = hma_p['projectCounties'].isna() & hma_p['county'].notna()

hma_p.loc[mask, 'projectCounties'] = hma_p.loc[mask, 'county'] + " County"

hma_p.isna().sum()
```

```python
# Fill categorical columns
hma_p['projectType'] = hma_p['projectType'].fillna('Unknown')
hma_p['recipient'] = hma_p['recipient'].fillna('Unknown')
hma_p['recipientTribalIndicator'] = hma_p['recipientTribalIndicator'].fillna(False)
hma_p['subrecipient'] = hma_p['subrecipient'].fillna('Unknown')
hma_p['county'] = hma_p['county'].fillna('Unknown')
hma_p['projectCounties'] = hma_p['projectCounties'].fillna('Unknown')

# Fill numeric columns
hma_p['projectAmount'] = hma_p['projectAmount'].fillna(0)
hma_p['costSharePercentage'] = hma_p['costSharePercentage'].fillna(75)
hma_p['benefitCostRatio'] = hma_p['benefitCostRatio'].fillna(hma_p['benefitCostRatio'].median())
hma_p['federalShareObligated'] = hma_p['federalShareObligated'].fillna(hma_p['federalShareObligated'].median())
hma_p['disasterNumber'] = hma_p['disasterNumber'].fillna(-1)

hma_p = hma_p.dropna(subset=['netValueBenefits'])

# hadling dates 
date_cols = [col for col in hma_p.columns if 'date' in col.lower()]

# Convert those columns to datetime format
for col in date_cols:
    hma_p[col] = pd.to_datetime(hma_p[col], errors='coerce')

print("Converted the following columns to datetime:")
print(date_cols)

hma_p['dateApproved'] = pd.to_datetime(hma_p['dateApproved'], errors='coerce')
hma_p['dateApproved'] = hma_p['dateApproved'].fillna(pd.Timestamp('1900-01-01'))
hma_p['dateApproved_missing'] = hma_p['dateApproved'] == pd.Timestamp('1900-01-01')

hma_p['dateClosed'] = pd.to_datetime(hma_p['dateClosed'], errors='coerce')
hma_p['project_open'] = hma_p['dateClosed'].isna()

hma_p.isna().sum()
```

```python
# Joining Black Homeowners Data 
bh = black_homeowners.copy()

# Split 'geo_name' into county and state
bh[['county', 'state']] = bh['geo_name'].str.replace(" County", "", regex=False)\
                                        .str.replace(" Parish", "", regex=False)\
                                        .str.split(", ", expand=True)

# Make sure formatting matches HMA dataset
bh['county'] = bh['county'].str.strip()
bh['state'] = bh['state'].str.strip()

bh.head()
```

```python
# Formatting features to join 

# Clean both datasets

hma_dis['state_x'] = hma_dis['state_x'].str.upper().str.strip()
hma_dis['county'] = hma_dis['county'].str.upper().str.strip().str.replace(" COUNTY", "").str.replace(" PARISH", "")

bh['state'] = bh['state'].str.upper().str.strip()
bh['county'] = bh['county'].str.upper().str.strip().str.replace(" COUNTY", "").str.replace(" PARISH", "")

```

```python

# Then merge
df_final = hma_dis.merge(
    bh,
    left_on=['state_x', 'county'],
    right_on=['state', 'county'],
    how='left'
)
```
